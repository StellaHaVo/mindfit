```python
#Nếu cột "parental_education_level" hoặc các cột khác chứa dữ liệu dạng chuỗi, hãy sử dụng Label Encoding (có thứ tự) hoặc One-Hot Encoding (0 có thứ tự):

from sklearn.preprocessing import LabelEncoder

encoder = LabelEncoder()
df["parental_education_level"] = encoder.fit_transform(df["parental_education_level"])

#kiểm tra lại
print(df.dtypes)  # Xem kiểu dữ liệu sau khi xử lý
```
